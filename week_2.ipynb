```python
import pandas as pd
```

```python
UniversitySalaries=pd.read_excel('/Users/melanyalegre/Downloads/University Salaries - Copy.xlsx')

print(UniversitySalaries.head())   # show first 5 rows
print(UniversitySalaries.dtypes)   # check column types
```

```python
import statsmodels.formula.api as sm
```

```python
model2=sm.ols('Salary~Age', data=UniversitySalaries).fit()
print(model2.summary())
```

```python
model3=sm.ols('Salary~Experience', data=UniversitySalaries).fit()
print(model3.summary())
```

```python
model4=sm.ols('Salary~Age+Experience', data=UniversitySalaries).fit()
print(model4.summary())
```

```python
import scipy.stats
scipy.stats.pearsonr(UniversitySalaries['Age'],
                     UniversitySalaries['Experience'])
                        
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor as vif 
from statsmodels.tools.tools import add_constant 

X=UniversitySalaries[['Age','Experience','Student_Approval']].dropna()
X=add_constant(X)

VIF_data=pd.DataFrame()
VIF_data["Variable"]= X.columns
VIF_data["VIF"]= [vif(X,i)
                   for i in range(len(X.columns))]
print(VIF_data)
```
